### 📘 Unsupervised Learning
##### What is Unsupervised Learning?

Unsupervised learning is a type of machine learning where:

There is no target variable (no labels)

The model tries to find hidden patterns or structures in data

It works with input features only

#### Unlike supervised learning (where we predict outputs), unsupervised learning focuses on discovering:

`Groups`

`Patterns`

`Relationships`

`Structures`

### Types of Unsupervised Learning

Unsupervised learning mainly has two major types:

`1. Clustering`

`2. Association`

### K-Means Clustering

**Concept:**  
An **unsupervised learning** algorithm (no target variable).  
It groups data points into **K clusters** based on similarity.

**Centroids:**  
The **center points** of each cluster.  
Each data point is assigned to the nearest centroid.

### Centroid Calculation in K-Means

####  Centroid Formula (1D Case)

The new centroid is calculated using the mean of all points in the cluster:

$$
\text{Centroid} = \frac{x_1 + x_2 + \dots + x_n}{n}
$$

Where:
- \( x_1, x_2, ..., x_n \) are data points
- \( n \) is the number of points in the cluster

## 📊 Centroid Formula (2D Case)

For 2D data points (x, y), the centroid has two coordinates:

$$
C_x = \frac{\sum x}{n}
$$

$$
C_y = \frac{\sum y}{n}
$$

So the centroid becomes:

$$
C = (C_x, C_y)
$$

# WCSS (Within-Cluster Sum of Squares)

**WCSS** measures how **compact** a cluster is.

- It is the **sum of squared distances** between each data point and its cluster centroid.
- Lower WCSS → points are **closer to their centroid**, indicating tight clusters.
- Higher WCSS → points are **spread out**, clusters are less compact.

---

## Formula for WCSS

For a cluster \(C_k\) with points \(x_i\) and centroid \( \mu_k \):

$$
WCSS = \sum_{k=1}^{K} \sum_{x_i \in C_k} \| x_i - \mu_k \|^2
$$

Where:  
- \(K\) = number of clusters  
- \(x_i\) = a point in cluster \(C_k\)  
- \(\mu_k\) = centroid of cluster \(C_k\)  
- \(\| x_i - \mu_k \|^2\) = squared Euclidean distance

# 📍 Elbow Method

The **Elbow Method** helps to find the **optimal number of clusters (K)** in K-Means.

- Idea: Compute **WCSS (Within-Cluster Sum of Squares)** for different values of K.
- Plot **K vs WCSS**.
- Look for the **“elbow point”**: where WCSS stops decreasing sharply.
  - Before the elbow → adding clusters reduces WCSS significantly
  - After the elbow → adding clusters doesn’t improve much

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
iris = load_iris()

df = pd.DataFrame()
df.tail()
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize the LabelEncoder
label_encoder = LabelEncoder()

# Fit and transform the data
df['Species'] = label_encoder.fit_transform(df['Species'])
print(label_encoder.classes_)
```

```python
x = df[["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]]
y = df[["Species"]]
```

```python
from sklearn.cluster import KMeans

cs = []
for i in range(1, 15):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0)
    kmeans.fit(x)
    cs.append(kmeans.inertia_)

plt.plot(range(1, 15), cs)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('CS')
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=0)

kmeans.fit(x)

labels = kmeans.labels_


y_pred = labels

for i in range(0,len(y_pred)):
    if y_pred[i] == 1:
        y_pred[i] = 0

    elif y_pred[i] == 0:
        y_pred[i] = 1

    else:
        y_pred[i] = 2


print('y_pred:',y_pred)
correct = sum(y_pred == list(y['Species']))
print(list(y['Species']))

print(f"The accuracy is: {correct/len(y_pred) * 100}%")
```

## Dataset: Mall Customer Segmentation (Annual Income vs Spending Score).

```python
import pandas as pd
from sklearn.cluster import KMeans


url = "https://gist.githubusercontent.com/GaneshSparkz/9dabfdeab9808d8e1b74f7fad4b91253/raw/Mall_Customers.csv"
df_mall = pd.read_csv(url)
df_mall.head()
```

```python
X = df_mall.iloc[:, [3, 4]].values
```

```python
df_mall.head()
```

```python
# 2. Elbow Method (Find optimal K)
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(range(1, 11), wcss, marker='o')
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# 3. Train K-Means with Optimal K (Assume K=5 based on plot)
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(X)
```
